## 신규 데이터

- 나이 : 35
- 성별 : 남
- 연소득 : 5000
- 근속연수 : 8
- 주거형태 : 자가
- 신용점수 : 720
- 기존대출건수 : 1
- 연간카드사용액 : 2400
- 부채비율 : 25.0
- 대출신청액 : 3000
- 대출목적 : 주택구입
- 상환방식 : 원리금균등
- 대출기간 : 36

```python
sample_input = {
    '나이' : 35,
    '성별' : '남', 
    '연소득' : 5000,
    '근속연수' : 8,
    '주거형태' : '자가',
    '신용점수' : 720,
    '기존대출건수' : 1,
    '연간카드사용액' : 2400,
    '부채비율' : 25.0,
    '대출신청액' : 3000,
    '대출목적' : '주택구입',
    '상환방식' : '원리금균등',
    '대출기간' : 36
}
```

```python
import joblib
import pandas as pd
```

```python
sample_df = pd.DataFrame([sample_input])
sample_df
```

```python
pipeline = joblib.load('models/loan_pipeline.pkl')
encoders = joblib.load('models/label_encoders.pkl')
features = joblib.load('models/feature_names.pkl')
```

```python
for col in ['성별', '주거형태', '대출목적', '상환방식']:
    sample_df[col] = encoders[col].transform(sample_df[col])
```

```python
sample_df
```

```python
pipeline.predict(sample_df)
```
